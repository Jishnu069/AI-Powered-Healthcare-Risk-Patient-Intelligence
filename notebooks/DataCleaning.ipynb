```python
weight_patterns = (
    known_weight
    .groupby(["age", "race", "gender", "weight"])
    .size()
    .reset_index(name="count")
)
```

```python
weight_patterns.head(20)
```

```python
weight_lookup = (
    weight_patterns
    .sort_values("count", ascending=False)
    .drop_duplicates(
        subset=["age", "race", "gender"]
    )
)
```

```python
weight_lookup.head(20)
```

```python
def fill_weight(age, race, gender):
    
    result = weight_lookup[
        (weight_lookup["age"] == age) &
        (weight_lookup["race"] == race) &
        (weight_lookup["gender"] == gender)
    ]
    
    if len(result) > 0:
        return result.iloc[0]["weight"]
    
    return np.nan
```

```python
df["weight_filled"] = df.apply(
    lambda row: fill_weight(
        row["age"],
        row["race"],
        row["gender"]
    ) if pd.isna(row["weight"]) else row["weight"],
    axis=1
)
```

```python
def convert_age(age):
    
    if pd.isna(age):
        return np.nan
    
    low, high = age[1:-1].split("-")
    
    return int((int(low) + int(high)) / 2)
```

```python
df["age"] = df["age"].apply(convert_age)
```

```python
df
```

```python
def convert_weight(weight):
    
    if pd.isna(weight):
        return np.nan
    
    if weight.startswith("["):
        low, high = weight[1:-1].split("-")
        return int((float(low) + float(high)) / 2)
    
    if weight == ">200":
        return 200
    
    return np.nan
```

```python
df["weight"] = df["weight"].apply(convert_weight)
```

```python
def categorize_icd9(code):

    if pd.isna(code):
        return "Unknown"

    code = str(code).strip()

    # V and E codes
    if code.startswith("V"):
        return "V-code"

    if code.startswith("E"):
        return "E-code"

    # Get the numeric part before the decimal
    try:
        number = float(code.split(".")[0])
    except ValueError:
        return "Unknown"

    # ICD-9 broad categories
    if 1 <= number <= 139:
        return "Infectious"

    elif 140 <= number <= 239:
        return "Neoplasms"

    elif 240 <= number <= 279:
        return "Endocrine"

    elif 280 <= number <= 289:
        return "Blood"

    elif 290 <= number <= 319:
        return "Mental"

    elif 320 <= number <= 389:
        return "Nervous"

    elif 390 <= number <= 459:
        return "Circulatory"

    elif 460 <= number <= 519:
        return "Respiratory"

    elif 520 <= number <= 579:
        return "Digestive"

    elif 580 <= number <= 629:
        return "Genitourinary"

    elif 630 <= number <= 679:
        return "Pregnancy"

    elif 680 <= number <= 709:
        return "Skin"

    elif 710 <= number <= 739:
        return "Musculoskeletal"

    elif 740 <= number <= 759:
        return "Congenital"

    elif 760 <= number <= 779:
        return "Perinatal"

    elif 780 <= number <= 799:
        return "Symptoms"

    elif 800 <= number <= 999:
        return "Injury"

    else:
        return "Other"
```

```python
test_codes = [
    "250.13",
    "401.9",
    "486",
    "E888",
    "V27"
]

for code in test_codes:
    print(code, "→", categorize_icd9(code))
```

### Encode categorical features

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)
```

### Identify numerical features

```python
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("Numerical features:")
print(numerical_features)
```

```python
print("\nNumber of numerical features:", len(numerical_features))
```

### Create the preprocessing setup

```python
from sklearn.compose import ColumnTransformer
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            encoder,
            categorical_features
        )
    ],
    remainder="passthrough"
)
```

### Train/Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Training target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True))
```

### Fit and transform the preprocessing

```python
X_train_processed = preprocessor.fit_transform(X_train)
```
